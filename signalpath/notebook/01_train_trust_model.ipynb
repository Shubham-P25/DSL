# SignalPath trust model training

This notebook trains only from a supplied labeled JSONL file; it never invents evaluation data.

```python
import json
import os
from pathlib import Path

data_path = os.getenv('TRUST_TRAIN_JSONL')
if not data_path:
    print('Set TRUST_TRAIN_JSONL to run training; no training performed.')
else:
    rows = [json.loads(line) for line in Path(data_path).read_text(encoding='utf-8').splitlines() if line.strip()]
    print(f'Loaded {len(rows)} labeled rows. Wire these rows into the selected sklearn/trust model before fitting.')
```
